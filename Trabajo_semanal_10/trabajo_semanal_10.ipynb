# Resolución

## 1) Impedancia de entrada del cuadripolo cargado

### <ins>Polinomio de Chebyshev para $n=3$</ins>

El cuadripolo $A$ debe ser un pasabajos Chebyshev de tercer orden, así que $n=3$. Trabajo con el circuito normalizado y las terminaciones $R_g=R_L=1\,\Omega$. Uso $\Omega$ para la frecuencia angular normalizada.

Llamo $C_n(\Omega)$ al polinomio de Chebyshev de orden $n$. Para el módulo cuadrado de la transmisión uso

$$|S_{21}(j\Omega)|^2=\frac{1}{1+\varepsilon^2[C_n(\Omega)]^2}.$$

Parto de $C_0(\Omega)=1$ y $C_1(\Omega)=\Omega$. Con la recurrencia $C_{k+1}=2\Omega C_k-C_{k-1}$ calculo primero $C_2$ y luego $C_3$.

$$
\begin{aligned}
C_2(\Omega)&=2\Omega C_1-C_0=2\Omega^2-1,\\
C_3(\Omega)&=2\Omega C_2-C_1
=2\Omega(2\Omega^2-1)-\Omega
=4\Omega^3-3\Omega.
\end{aligned}
$$

Adopto un ripple de $\alpha_p=0{,}5\,\mathrm{dB}$. El factor $\varepsilon$ que entra en la expresión anterior queda

$$
\varepsilon=\sqrt{10^{\alpha_p/10}-1}
=\sqrt{10^{0{,}5/10}-1}\approx0{,}3493.
$$

### <ins>Polos del prototipo y cálculo de $S_{21}$</ins>

Para calcular los polos uso $\mu$ (el parámetro $a$ del manuscrito). Primero escribo la fórmula en función del orden y después sustituyo $n=3$.

$$
\mu=\frac{1}{n}\operatorname{asinh}\!\left(\frac{1}{\varepsilon}\right)
\quad\xrightarrow{n=3}\quad
\mu=\frac{1}{3}\operatorname{asinh}\!\left(\frac{1}{\varepsilon}\right)
\approx0{,}5914.
$$

A cada polo le corresponde un ángulo $\theta_k=(2k-1)\pi/(2n)$. Como $2n=6$, los ángulos son $\pi/6$, $\pi/2$ y $5\pi/6$. Al sustituir cada uno en la expresión de los polos resulta

$$
p_k=-\sinh(\mu)\sin\theta_k+j\cosh(\mu)\cos\theta_k,
\qquad
\begin{aligned}
p_1&\approx-0{,}3132+j1{,}0219,\\
p_2&\approx-0{,}6265,\\
p_3&\approx-0{,}3132-j1{,}0219.
\end{aligned}
$$

El polo real aporta el factor $s-p_2$. Para los otros dos multiplico $(s-p_1)(s-p_3)$. Como son conjugados, el producto da un polinomio de segundo orden con coeficientes reales.

$$
G(s)\approx(s+0{,}6265)(s^2+0{,}6264s+1{,}1424)
\approx s^3+1{,}2529s^2+1{,}5349s+0{,}7157.
$$

Con el denominador así armado, la transferencia de transmisión queda

$$
\boxed{S_{21}(s)=T(s)=\frac{K}{G(s)}}
\approx\frac{0{,}7157}{s^3+1{,}2529s^2+1{,}5349s+0{,}7157}.
$$

### <ins>Obtención de $S_{11}$</ins>

Como el cuadripolo no disipa potencia, $|S_{21}|^2+|S_{11}|^2=1$. Con $B=\varepsilon^2(4\Omega^3-3\Omega)^2$, la cuenta queda

$$
|S_{21}|^2=\frac{1}{1+B},\qquad
|S_{11}|^2=1-\frac{1}{1+B}=\frac{B}{1+B}.
$$

Las dos fracciones tienen el mismo denominador. Por eso $|S_{11}|^2=B|S_{21}|^2$ y, al sacar raíz, resulta

$$|S_{11}|=\varepsilon\,|4\Omega^3-3\Omega|\,|S_{21}|.$$

Para escribir el polinomio en función de $s$, despejo $\Omega=-js$ a partir de $s=j\Omega$.

$$
4(-js)^3-3(-js)=4js^3+3js=j(4s^3+3s).
$$

El factor $j$ tiene módulo uno. Así, elijo $S_{11}(s)=\varepsilon(4s^3+3s)S_{21}(s)$, que da el módulo calculado. Para simplificar uso el valor exacto de $K$ del prototipo.

$$K=\frac{1}{2^{n-1}\varepsilon}=\frac{1}{4\varepsilon}\approx0{,}7157.$$

Al reemplazar $S_{21}=K/G$ se cancela $\varepsilon$.

$$
\boxed{
S_{11}(s)=\frac{\varepsilon K(4s^3+3s)}{G(s)}
=\frac{s^3+\frac34s}{G(s)}
=\frac{F(s)}{G(s)}
}.
$$

### <ins>Impedancia de entrada $Z_1$</ins>

Con la carga de $1\,\Omega$ conectada, el coeficiente de reflexión en la entrada es $S_{11}=(Z_1-1)/(Z_1+1)$. Al multiplicar por $Z_1+1$ y agrupar los términos con $Z_1$, queda $Z_1(1-S_{11})=1+S_{11}$. Sustituyo $S_{11}=F/G$.

$$
\boxed{Z_1(s)=\frac{1+S_{11}}{1-S_{11}}
=\frac{1+F/G}{1-F/G}=\frac{G+F}{G-F}}.
$$

El término $s^3$ se suma arriba y se cancela abajo. Con los coeficientes usados en las divisiones,

$$
\boxed{
Z_1(s)\approx
\frac{2s^3+1{,}252913s^2+2{,}284895s+0{,}715694}
{1{,}252913s^2+0{,}784895s+0{,}715694}
}.
$$

## 2) Síntesis del cuadripolo escalera

### <ins>Primera remoción en infinito</ins>

Divido el numerador de $Z_1$ por su denominador $D_{Z_1}(s)$. El cociente sale de dividir los términos de mayor grado. Después multiplico por el divisor y resto. En las cuentas muestro los coeficientes redondeados.

$$
\begin{array}{rrrrr|l}
 & 2s^3 & +1{,}25291297s^2 & +2{,}28489546s & +0{,}71569379 &
\underline{1{,}25291297s^2+0{,}78489546s+0{,}71569379}\\
-& 2s^3 & +1{,}25291297s^2 & +1{,}14244773s & &
1{,}59628s\\
\hline
 & 0 & 0 & 1{,}14244773s & +0{,}71569379 &
\end{array}
$$

El **cociente** es $1{,}59628s$. Como sale de una impedancia, es el término $sL_{1,n}$ de la primera bobina serie. El **resto polinómico** es $R_1(s)\approx1{,}14244773s+0{,}71569379$. La impedancia que queda es $R_1/D_{Z_1}$. La invierto para pasar a la admitancia $Y_2=D_{Z_1}/R_1$.

### <ins>Segunda remoción en infinito</ins>

En $Y_2$ se invierten los papeles de los polinomios. Ahora divido $D_{Z_1}(s)$ por el resto $R_1(s)$ de la cuenta anterior.

$$
\begin{array}{rrrr|l}
 & 1{,}25291297s^2 & +0{,}78489546s & +0{,}71569379 &
\underline{1{,}14244773s+0{,}71569379}\\
-& 1{,}25291297s^2 & +0{,}78489546s & &
1{,}09669s\\
\hline
 & 0 & 0 & 0{,}71569379 &
\end{array}
$$

El **cociente** es $1{,}09669s$. Como estoy dividiendo una admitancia, es el término $sC_{\mathrm{norm}}$ del capacitor en derivación. El **resto polinómico** es $R_2\approx0{,}71569379$. Queda la admitancia $R_2/R_1$. Al invertirla obtengo $Z_3=R_1/R_2$.

### <ins>Tercera remoción en infinito</ins>

Divido $R_1(s)$ por $R_2$ y retiro solo el término en $s$, como en la cuenta del manuscrito.

$$
\begin{array}{rrr|l}
 & 1{,}14244773s & +0{,}71569379 & \underline{0{,}71569379}\\
-& 1{,}14244773s & & 1{,}59628s\\
\hline
 & 0 & 0{,}71569379 &
\end{array}
$$

El **cociente extraído** es $1{,}59628s$, que corresponde a la segunda bobina serie $sL_{2,n}$. El **resto polinómico** es $R_3\approx0{,}71569379$.

### <ins>Carga al final de la escalera</ins>

Invierto la impedancia que queda, $R_3/R_2$, y hago la última división en admitancia.

$$
\begin{array}{rr|l}
 & 0{,}71569379 & \underline{0{,}71569379}\\
-& 0{,}71569379 & 1\\
\hline
 & 0 &
\end{array}
$$

El **cociente** es $1$ y el **resto** es $0$. La admitancia final es $1\,\mathrm{S}$, que corresponde a la carga $R_L=1\,\Omega$.

### <ins>Conversión de las frecuencias de los cursores</ins>

LTspice muestra la frecuencia $f$ en Hz o mHz. Para ubicar cada lectura en la respuesta del prototipo, primero paso a frecuencia angular con $\omega=2\pi f$ y después divido por la frecuencia angular del borde de paso. Como el circuito está normalizado, $\omega_p=1\,\mathrm{rad/s}$.

$$\boxed{\Omega=\frac{\omega}{\omega_p}=\frac{2\pi f}{\omega_p}},\qquad
\boxed{f=\frac{\Omega\,\omega_p}{2\pi}}.$$

Por ejemplo, el cursor del punto medio de la banda de paso marca $79{,}582968\,\mathrm{mHz}$. La cuenta para esa lectura es

$$
\begin{aligned}
f&=79{,}582968\,\mathrm{mHz}=0{,}079582968\,\mathrm{Hz},\\
\omega&=2\pi f\approx0{,}500035\,\mathrm{rad/s},\\
\Omega&=\frac{0{,}500035\,\mathrm{rad/s}}{1\,\mathrm{rad/s}}\approx0{,}500035.
\end{aligned}
$$

El resultado queda cerca de $\Omega=0{,}5$, la mitad del tramo positivo de la banda de paso, entre $\Omega=0$ y $\Omega=1$. En la tabla aplico la misma conversión al resto de las lecturas. La columna de referencia indica el punto teórico que busqué con cada cursor.

| Punto | $\Omega$ de referencia | $f$ del cursor (mHz) | $\Omega$ del cursor | Módulo de $S_{11}$ (dB) |
|:--|--:|--:|--:|--:|
| Punto medio del tramo positivo de la banda de paso | $0{,}5$ | 79,582968 | 0,500035 | −9,635745 |
| Borde de paso | $1$ | 159,15494 | 1,000000 | −9,635745 |
| Transición, cerca del punto de media potencia | $1{,}167485$ | 185,70352 | 1,166810 | −3,024022 |
| Referencia en la banda detenida | $3$ | 477,33154 | 2,999163 | −0,003636 |
| Nulo de reflexión | $\sqrt{3}/2\approx0{,}866025$ | 137,95961 | 0,866826 | −54,05885 |

Si $M_{11}$ es la lectura del módulo en dB, la fracción de potencia reflejada es $|S_{11}|^2=10^{M_{11}/10}$.

### <ins>Banda de paso y frecuencia de corte</ins>

En el tramo positivo, el punto medio está cerca de $79{,}58\,\mathrm{mHz}$, la mitad de la frecuencia de corte. En ese punto y en el borde de paso, a $159{,}15\,\mathrm{mHz}$, la lectura da $-9{,}64\,\mathrm{dB}$ al redondear. En ambos casos se refleja cerca del $10{,}87\,\%$ de la potencia incidente.

Dentro de la banda detenida tomo $\Omega=3$ como referencia. La lectura más próxima que usé es $477{,}33\,\mathrm{mHz}$. Ahí el módulo medido está muy cerca de $0\,\mathrm{dB}$ y se refleja cerca del $99{,}92\,\%$ de la potencia incidente. Esto concuerda con la caída de $S_{21}$ que se observa en el inciso anterior.

En Hz, ese cero corresponde a $137{,}83\,\mathrm{mHz}$. La lectura queda cerca de ese punto de adaptación, donde $Z_1=1\,\Omega$ y la potencia incidente se transmite a la carga.

## 5) Desnormalización a $1\,\mathrm{MHz}$ y $50\,\Omega$

La frecuencia angular pedida es $\omega_c=2\pi\cdot10^6\,\mathrm{rad/s}$. En Hz equivale a $f_c=\omega_c/(2\pi)=1\,\mathrm{MHz}$. Para el generador de $50\,\Omega$ uso $R_0=50\,\Omega$ como norma de impedancia.

Aplico las reglas de desnormalización a las resistencias, las bobinas y el capacitor.

$$R=R_0R_n,\qquad L=\frac{R_0L_n}{\omega_c},\qquad C=\frac{C_{\mathrm{norm}}}{R_0\omega_c}.$$

Las resistencias normalizadas valen uno, así que $R_g=R_L=50\,\Omega$. Para los demás componentes sustituyo los valores que salieron de las remociones.

$$
\begin{aligned}
L_1=L_2&=\frac{50\cdot1{,}59628}{2\pi\cdot10^6}\approx12{,}7028\,\mu\mathrm{H},\\
C&=\frac{1{,}09669}{50\cdot2\pi\cdot10^6}\approx3{,}49088\,\mathrm{nF}.
\end{aligned}
$$
